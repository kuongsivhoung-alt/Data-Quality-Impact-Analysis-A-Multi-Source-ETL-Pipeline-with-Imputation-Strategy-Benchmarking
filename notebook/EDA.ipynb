````python

#function for missing value and percentage
def missing(df,table_name):
    mis = df.isna().sum()
    mis_pct = df.isna().mean()*100

    result = pd.DataFrame({
        'table': table_name,
        'column':df.columns,
        'missing_count':mis.values,
        'missing_percentage': mis_pct.values
    })
    return result[result['missing_count']>0]
````

````python
missing_results = []

for table_name, df in table.items():

    result = missing(df, table_name)

    missing_results.append(result)
````

````python
missing_summary = pd.concat(missing_results, ignore_index=True)
print(missing_summary)
````

### 5. Missingness Mechanism Analysis

**Test 1: review_score ↔ message_missing**

````python
# review score and missing message
pd.crosstab(
    order_reviews["review_score"],
    order_reviews["message_missing"]
)
````

````python
# Chi-square test of independence

from scipy.stats import chi2_contingency

table = pd.crosstab(
    order_reviews["review_score"],
    order_reviews["message_missing"]
)

chi2, p, dof, expected = chi2_contingency(table)

print(f"Chi-square: {chi2:.2f}")
print(f"p-value: {p:.2e}")
print(f"Degrees of freedom: {dof}")
print(f"Minimum expected frequency: {expected.min():.2f}")
````

The chi-square test showed a statistically significant association between review_score and message_missing (χ²(4) = 8307.00, p < 0.001). Therefore, we reject the null hypothesis of independence.

The expected frequencies are sufficiently large, so the chi-square test is appropriate.
